```python
import numpy as np
import pandas as pd
# -----------------------------
# 1. DATASET
# -----------------------------
data = {
    "Age": [22, 25, 28, 35, 40, 45, 50, 23, 30, 38],
    "Visits": [2, 5, 6, 8, 7, 3, 2, 6, 7, 9],
    "Time_Spent": [5, 15, 20, 35, 30, 10, 8, 25, 28, 40],
    "Previous_Purchases": [0, 1, 2, 4, 3, 1, 0, 2, 3, 5],
    "Bought": [0, 0, 1, 1, 1, 0, 0, 1, 1, 1]
}

df = pd.DataFrame(data)
print(df)
# -----------------------------
# 2. TREE 1
# -----------------------------
def tree1(visits):
    if visits > 5:
        return 1
    else:
        return 0
# -----------------------------
# 3. TREE 2
# -----------------------------
def tree2(time_spent):

    if time_spent > 20:
        return 1
    else:
        return 0
# -----------------------------
# 4. TREE 3
# -----------------------------

def tree3(previous_purchases):

    if previous_purchases > 1:
        return 1
    else:
        return 0


# -----------------------------
# 5. RANDOM FOREST
# -----------------------------

def random_forest(visits, time_spent, previous_purchases):

    prediction1 = tree1(visits)
    prediction2 = tree2(time_spent)
    prediction3 = tree3(previous_purchases)
    predictions = [
        prediction1,
        prediction2,
        prediction3
    ]
    print("\nTree Predictions:")
    print("Tree 1:", prediction1)
    print("Tree 2:", prediction2)
    print("Tree 3:", prediction3)

    # Majority voting
    final_prediction = max( #max value
        set(predictions),
        key=predictions.count
    )
    return final_prediction


# -----------------------------
# 6. NEW CUSTOMER
# -----------------------------

result = random_forest(
    visits=7,
    time_spent=25,
    previous_purchases=3
)


# -----------------------------
# 7. FINAL RESULT
# -----------------------------

print("\nFinal Prediction:")

if result == 1:
    print("Customer is likely to BUY 🛒")
else:
    print("Customer is unlikely to BUY ❌")
```

```python
import pandas as pd
import numpy as np

# ==========================
# LOAD DATA
# ==========================

df = pd.read_csv("titanic.csv")

# Keep required columns
df = df[['Pclass', 'Sex', 'Age', 'Fare', 'Survived']]

# Fill missing age values
df['Age'] = df['Age'].fillna(df['Age'].median())

# Convert Sex to numeric
df['Sex'] = df['Sex'].map({
    'female': 0,
    'male': 1
})

# ==========================
# SIMPLE DECISION TREES
# ==========================

# Tree 1 -> Based on Gender

def tree1(sex):

    if sex == 0:
        return 1
    else:
        return 0


# Tree 2 -> Based on Passenger Class

def tree2(pclass):

    if pclass <= 2:
        return 1
    else:
        return 0


# Tree 3 -> Based on Age

def tree3(age):

    if age < 15:
        return 1
    else:
        return 0


# Tree 4 -> Based on Fare

def tree4(fare):

    if fare > 30:
        return 1
    else:
        return 0


# ==========================
# RANDOM FOREST
# ==========================

def random_forest(passenger):

    votes = []

    votes.append(tree1(passenger['Sex']))
    votes.append(tree2(passenger['Pclass']))
    votes.append(tree3(passenger['Age']))
    votes.append(tree4(passenger['Fare']))

    prediction = max(
        set(votes),
        key=votes.count
    )

    return prediction


# ==========================
# TEST ON FIRST 10 ROWS
# ==========================

correct = 0

for index, row in df.head(10).iterrows():

    actual = row['Survived']

    predicted = random_forest(row)

    print(
        f"Passenger {index} | Actual={actual} | Predicted={predicted}"
    )

    if actual == predicted:
        correct += 1


accuracy = correct / 10

print("\nAccuracy:", round(accuracy, 2))
# ==========================
# PREDICT NEW PASSENGER
# ==========================
new_passenger = {
    'Pclass': 1,
    'Sex': 0,
    'Age': 25,
    'Fare': 100
}

prediction = random_forest(new_passenger)

print("\nNew Passenger Prediction:")

if prediction == 1:
    print("Survived")
else:
    print("Did Not Survive")
```
