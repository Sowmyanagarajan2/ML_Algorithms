```python
# ============================================================
# LINEAR REGRESSION FROM SCRATCH
# Problem: Predict salary based on years of experience
# Libraries: NumPy + Pandas
# Experience
#     ↓
# Find pattern
#     ↓
# Calculate slope
#     ↓
# Calculate intercept
#     ↓
# y = mx + b
#     ↓
# Predict Salary
# ============================================================

import numpy as np
import pandas as pd
# ------------------------------------------------------------
# STEP 1: Create the dataset
# ------------------------------------------------------------

data = {
    "Experience": [1, 2, 3, 4, 5],
    "Salary": [30, 35, 40, 50, 55]
}

# Convert dictionary into a Pandas DataFrame
df = pd.DataFrame(data)

# print("Dataset:")
# print(df)


# ------------------------------------------------------------
# STEP 2: Separate input (X) and output (y)
# ------------------------------------------------------------

# X = Independent variable / Input
X = df["Experience"].values
# y = Dependent variable / Output
y = df["Salary"].values
# print("\nX (Experience):")
# print(X)
# print("\ny (Salary):")
# print(y)
# ------------------------------------------------------------
# STEP 3: Calculate the mean
# ------------------------------------------------------------
# Mean of X
x_mean = np.mean(X)
# Mean of y
y_mean = np.mean(y)
print("\nMean of X:", x_mean)
print("Mean of y:", y_mean)
# ------------------------------------------------------------
# STEP 4: Calculate the slope (m)
# ------------------------------------------------------------

# Formula:
#y = mx + b -> final prediction formula but to know the slope we are working on this
#             Σ((x - x_mean)(y - y_mean))
# m = --------------------------------------------
#             Σ((x - x_mean)^2)
# Slope tells us how much y changes when x increases by 1 unit.
numerator = np.sum((X - x_mean) * (y - y_mean))
denominator = np.sum((X - x_mean) ** 2)
m = numerator / denominator
print("\nSlope (m):", m)

# ------------------------------------------------------------
# STEP 5: Calculate the intercept (b)
# ------------------------------------------------------------
# Formula:
#
# b = y_mean - (m * x_mean)

b = y_mean - (m * x_mean)

print("Intercept (b):", b)


# ------------------------------------------------------------
# STEP 6: Create the Linear Regression equation
# ------------------------------------------------------------

print("\nLinear Regression Equation:")

print(
    "Salary =", m,
    "* Experience +", b
)


# ------------------------------------------------------------
# STEP 7: Predict salary for a new employee
# ------------------------------------------------------------

new_experience = 7

# Formula:
#
# y = mx + b

prediction = (m * new_experience) + b
# (1*7)+30

print("\nNew Employee Experience:", new_experience)

print("Predicted Salary:", prediction)
```

```python
# ============================================================
# LOGISTIC REGRESSION FROM SCRATCH
# Problem: Predict whether a student will Pass or Fail
# Libraries: NumPy + Pandas
# ============================================================

import numpy as np
import pandas as pd
# ------------------------------------------------------------
# STEP 1: Create dataset
# ------------------------------------------------------------
data = {
    "StudyHours": [1, 2, 3, 4, 5, 6],
    "Pass": [0, 0, 0, 1, 1, 1] # 0 fail,pass 1
}
df = pd.DataFrame(data)
# ------------------------------------------------------------
# STEP 2: Separate X and y
# ------------------------------------------------------------
X = df["StudyHours"].values
y = df["Pass"].values
# ------------------------------------------------------------
# STEP 3: Initialize weight and bias
# ------------------------------------------------------------
# Weight controls the importance of X
weight = 0.0
# Bias shifts the prediction
bias = 0.0
# ------------------------------------------------------------
# STEP 4: Set learning parameters
# ------------------------------------------------------------
learning_rate = 0.1
# Number of times the model learns from the data
epochs = 1000
# ------------------------------------------------------------
# STEP 5: Sigmoid function
# ------------------------------------------------------------

# Sigmoid converts any number into a value between 0 and 1.
#
# Formula:
#
# sigmoid(z) = 1 / (1 + e^(-z))
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# ------------------------------------------------------------
# STEP 6: Train the model
# ------------------------------------------------------------

for epoch in range(epochs):

    # Calculate linear score
    #
    # z = wx + b

    z = weight * X + bias

    # Convert score into probability

    probability = sigmoid(z)

    # --------------------------------------------------------
    # Calculate gradients
    # --------------------------------------------------------

    # Difference between prediction and actual value

    error = probability - y
    # Gradient for weight
    weight_gradient = np.mean(error * X)
    # Gradient for bias
    bias_gradient = np.mean(error)
    # --------------------------------------------------------
    # Update weight and bias
    # --------------------------------------------------------
    weight = weight - learning_rate * weight_gradient
    bias = bias - learning_rate * bias_gradient
# ------------------------------------------------------------
# STEP 7: Display learned values
# ------------------------------------------------------------
print("\nLearned Weight:", weight)
print("Learned Bias:", bias)
# ------------------------------------------------------------
# STEP 8: Create prediction function
# ------------------------------------------------------------

def predict(hours):
    # Calculate linear score
    z = weight * hours + bias
    # Convert score into probability
    probability = sigmoid(z)
    # Apply threshold
    if probability >= 0.5:
        prediction = 1
    else:
        prediction = 0
    return probability, prediction
# ------------------------------------------------------------
# STEP 9: Test with a new student
# ------------------------------------------------------------
hours = 5

probability, prediction = predict(hours)
print("\nStudy Hours:", hours)
print("Pass Probability:", probability)
print("Prediction:", prediction)


# ------------------------------------------------------------
# STEP 10: Display human-readable result
# ------------------------------------------------------------

if prediction == 1:

    print("Result: PASS")
else:

    print("Result: FAIL")
```

```python
# ============================================================
# NAIVE BAYES FROM SCRATCH
# Problem: Detect Spam messages
# Libraries: NumPy + Pandas
# ============================================================

import numpy as np
import pandas as pd


# ------------------------------------------------------------
# STEP 1: Create dataset
# ------------------------------------------------------------

data = {
    "Message": [
        "free money",
        "win prize",
        "free prize",
        "meeting today",
        "project meeting",
        "team meeting"
    ],

    "Class": [
        "Spam",
        "Spam",
        "Spam",
        "Not Spam",
        "Not Spam",
        "Not Spam"
    ]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)


# ------------------------------------------------------------
# STEP 2: Calculate prior probabilities
# ------------------------------------------------------------

# Count Spam messages

spam_count = np.sum(
    df["Class"] == "Spam"
)


# Count Not Spam messages

not_spam_count = np.sum(
    df["Class"] == "Not Spam"
)


# Total number of messages

total_messages = len(df)


# Prior probability of Spam

p_spam = spam_count / total_messages


# Prior probability of Not Spam

p_not_spam = not_spam_count / total_messages


print("\nPrior Probabilities:")

print("P(Spam):", p_spam)

print("P(Not Spam):", p_not_spam)


# ------------------------------------------------------------
# STEP 3: Create vocabulary
# ------------------------------------------------------------

# Combine all messages into one string

all_text = " ".join(df["Message"])


# Split the text into words

words = all_text.lower().split()


# Remove duplicate words

vocabulary = np.unique(words)


print("\nVocabulary:")

print(vocabulary)


# ------------------------------------------------------------
# STEP 4: Calculate word probabilities
# ------------------------------------------------------------

def word_probability(word, message_class):

    # Get messages belonging to the required class

    messages = df[
        df["Class"] == message_class
    ]["Message"]


    # Count how many messages contain the word

    count = 0

    for message in messages:

        if word in message.lower().split():

            count += 1


    # Number of messages in that class

    class_count = len(messages)


    # Laplace smoothing
    #
    # +1 prevents zero probability

    probability = (
        count + 1
    ) / (
        class_count + 2
    )

    return probability


# ------------------------------------------------------------
# STEP 5: Create prediction function
# ------------------------------------------------------------

def predict(message):

    # Convert message into individual words

    words = message.lower().split()


    # Start with prior probabilities

    spam_score = p_spam

    not_spam_score = p_not_spam


    # Look at every word

    for word in words:

        # Probability of word given Spam

        p_word_spam = word_probability(
            word,
            "Spam"
        )


        # Probability of word given Not Spam

        p_word_not_spam = word_probability(
            word,
            "Not Spam"
        )


        # Multiply probabilities

        spam_score *= p_word_spam

        not_spam_score *= p_word_not_spam


    # Compare both scores

    if spam_score > not_spam_score:

        result = "Spam"

    else:

        result = "Not Spam"


    return result, spam_score, not_spam_score


# ------------------------------------------------------------
# STEP 6: Test a new message
# ------------------------------------------------------------

new_message = "free money"

result, spam_score, not_spam_score = predict(
    new_message
)


print("\nNew Message:")

print(new_message)


print("\nSpam Score:")

print(spam_score)


print("\nNot Spam Score:")

print(not_spam_score)


print("\nPrediction:")

print(result)
```

```python
# ============================================================
# K-NEAREST NEIGHBORS (KNN) FROM SCRATCH
# Problem: Predict customer category
# Libraries: NumPy + Pandas
# ============================================================

import numpy as np
import pandas as pd


# ------------------------------------------------------------
# STEP 1: Create dataset
# ------------------------------------------------------------

data = {
    "Age": [20, 22, 25, 28, 30, 35],

    "Income": [20, 25, 30, 35, 40, 45],

    "Class": [
        "Low",
        "Low",
        "High",
        "High",
        "High",
        "High"
    ]
}

df = pd.DataFrame(data)

print("Dataset:")

print(df)


# ------------------------------------------------------------
# STEP 2: Separate features and target
# ------------------------------------------------------------

# Get Age and Income as NumPy array

X = df[
    ["Age", "Income"]
].values


# Get class labels

y = df["Class"].values


# ------------------------------------------------------------
# STEP 3: Create a new customer
# ------------------------------------------------------------

new_customer = np.array(
    [26, 32]
)

print("\nNew Customer:")

print("Age:", new_customer[0])

print("Income:", new_customer[1])


# ------------------------------------------------------------
# STEP 4: Calculate distance
# ------------------------------------------------------------

# Euclidean Distance:
#
# distance =
# √((x2-x1)^2 + (y2-y1)^2)
#
# We calculate distance from the new customer
# to every existing customer.

distances = np.sqrt(
    np.sum(
        (X - new_customer) ** 2,
        axis=1
    )
)


# ------------------------------------------------------------
# STEP 5: Add distances to DataFrame
# ------------------------------------------------------------

df["Distance"] = distances

print("\nDistance from New Customer:")

print(df)


# ------------------------------------------------------------
# STEP 6: Select K nearest neighbors
# ------------------------------------------------------------

# K means number of neighbors we want to check

K = 3


# argsort gives indexes in ascending order

nearest_indices = np.argsort(
    distances
)[:K]


print("\nNearest Customer Indexes:")

print(nearest_indices)


# ------------------------------------------------------------
# STEP 7: Get the classes of nearest neighbors
# ------------------------------------------------------------

nearest_classes = y[
    nearest_indices
]


print("\nNearest Neighbor Classes:")

print(nearest_classes)


# ------------------------------------------------------------
# STEP 8: Count votes
# ------------------------------------------------------------

values, counts = np.unique(
    nearest_classes,
    return_counts=True
)


print("\nClasses:")

print(values)


print("\nVotes:")

print(counts)


# ------------------------------------------------------------
# STEP 9: Find class with highest votes
# ------------------------------------------------------------

winner_index = np.argmax(
    counts
)


prediction = values[
    winner_index
]


# ------------------------------------------------------------
# STEP 10: Display final prediction
# ------------------------------------------------------------

print("\nFinal Prediction:")

print(prediction)
```
