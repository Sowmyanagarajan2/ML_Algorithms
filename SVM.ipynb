SVM is normally supervised because it learns from labeled data. One-Class SVM is a special version that can learn what normal data looks like and detect unusual data without requiring normal-vs-abnormal class labels

*   SVM tries to draw the best line between the two groups.
*   👉 Decision Boundary
*   "Where should I draw the line so that the two groups are separated as well as possible?"
------------------------------------------------------------------------
Types of SVM: Linear & Non Linear

**SVM is not just "draw a line."**

It does three important things:

Finds a decision boundary
Maximizes the margin

Advantages

1.High-Dimensional Performance: SVM excels in high-dimensional spaces, making it

2.suitable for image classification and gene expression analysis.

3.Nonlinear Capability: Utilizing kernel functions like RBF and polynomial SVM effectively handles nonlinear relationships.

4.utlier Resilience: SVM can handle outliers by allowing some misclassifications using a soft margin.

5.Binary and Multiclass Support: SVM is effective for both binary classification and multiclass classification suitable for applications in text classification.

6.Memory Efficiency: It focuses on support vectors making it memory efficient compared to other algorithms

**How its working?**

**Features → Decision Formula → Decision Value → Class → Margin → Correct/Incorrect.**

🟢 Decision Boundary

The line separating the classes.

🟢 Margin

The safe distance around the decision boundary.

🟢 Support Vectors

The data points closest to the boundary.

🟢 SVM

Finds a boundary with a maximum margin.

````python
# ============================================================
# SUPPORT VECTOR MACHINE (SVM) - SIMPLE CALCULATION
# ============================================================

# REAL-WORLD EXAMPLE:
# We want to classify a mobile phone as:
#
# -1 = Budget Phone
# +1 = Premium Phone
#
# Features:
# x1 = RAM (GB)
# x2 = Price (₹ Thousand)


# ============================================================
# STEP 1: GIVEN DATA
# ============================================================

RAM = 6
Price = 28

# SVM learned weights
w1 = 2
w2 = 0.5

# Bias
b = -20


print("========== PHONE DETAILS ==========")
print("RAM   =", RAM, "GB")
print("Price =", Price, "₹ Thousand")


# ============================================================
# STEP 2: SVM DECISION FORMULA
# ============================================================

# Formula:
#
# Decision = w1*x1 + w2*x2 + b


print("\n========== SVM FORMULA ==========")

print("Formula:")
print("Decision = w1*x1 + w2*x2 + b")


# ============================================================
# STEP 3: SUBSTITUTE THE VALUES
# ============================================================

print("\n========== SUBSTITUTE VALUES ==========")

print(
    "Decision =",
    w1, "*", RAM,
    "+",
    w2, "*", Price,
    "+",
    b
)


# ============================================================
# STEP 4: CALCULATE
# ============================================================

decision = (w1 * RAM) + (w2 * Price) + b

print("\n========== CALCULATION ==========")

print(
    "Decision =",
    w1 * RAM,
    "+",
    w2 * Price,
    "+",
    b
)

print("Decision =", decision)


# ============================================================
# STEP 5: MAKE PREDICTION
# ============================================================

print("\n========== PREDICTION ==========")

if decision > 0:
#y=1
    print("Decision > 0")
    print("Prediction = +1")
    print("Phone Type = PREMIUM")

else:
#y=0
    print("Decision < 0")
    print("Prediction = -1")
    print("Phone Type = BUDGET")


# ============================================================
# STEP 6: SVM MARGIN
# ============================================================

# Formula:
#0 or 1
# Margin = y * (w1*x1 + w2*x2 + b)
#y is op of prediction
# Premium = +1
# Budget  = -1

y = 1       # Actual class: Premium
margin = y * decision


print("\n========== MARGIN CALCULATION ==========")

print("Formula:")
print("Margin = y * (w1*x1 + w2*x2 + b)")
print("\nSubstitution:")
print(
    "Margin =",
    y,
    "*",
    "(",
    w1, "*", RAM,
    "+",
    w2, "*", Price,
    "+",
    b,
    ")"
)
print("\nMargin =", margin)
# ============================================================
# STEP 7: CHECK THE MARGIN
# ============================================================

print("\n========== MARGIN CHECK ==========")

if margin >= 1:
    print("Margin >= 1")
    print("Correctly classified!")
    print("Point is outside the margin.")

else:
    print("Margin < 1")
    print("Point is inside the margin or incorrectly classified.")
# ============================================================
# STEP 8: DECISION BOUNDARY
# ============================================================

# Decision Boundary Formula:
#
# w1*x1 + w2*x2 + b = 0

print("\n========== DECISION BOUNDARY ==========")

print("Formula:")
print("w1*x1 + w2*x2 + b = 0")

print("\nOur Boundary:")

print(
    w1,
    "* RAM +",
    w2,
    "* Price +",
    b,
    "= 0"
)

print("\nTherefore:")

print("2 * RAM + 0.5 * Price - 20 = 0")


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n========================================")
print("              SVM SUMMARY")
print("========================================")

print("RAM                  :", RAM, "GB")
print("Price                :", Price, "₹ Thousand")
print("Decision Value       :", decision)
print("Prediction           :", "Premium" if decision > 0 else "Budget")
print("Margin               :", margin)
print("Correctly Classified :", "YES" if margin >= 1 else "NO")

print("\nKey Formula:")
print("Decision = w1*x1 + w2*x2 + b")

print("\nMargin Formula:")
print("Margin = y * (w1*x1 + w2*x2 + b)")

print("\nDecision Boundary:")
print("w1*x1 + w2*x2 + b = 0")
````
